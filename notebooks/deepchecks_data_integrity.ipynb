### Data location

This notebook was developed in Google Colab. Update `project_dir` to the location of `bank-additional-full.csv` in your own environment.

```python
# Colab configuration - update this path if the project is stored elsewhere
project_dir = Path('/content/drive/My Drive/AITesting_Proj')
data_path = project_dir/"bank-additional-full.csv"

bank_data = pd.read_csv(data_path, sep = ';')

categorical_features = ['job', 'marital', 'education', 'default', 'housing',
                        'loan', 'contact', 'month', 'day_of_week', 'poutcome']

ds = Dataset(bank_data, label = "y", cat_features=categorical_features)

integrity_suite = data_integrity()
bank_suite_result = integrity_suite.run(ds)
bank_suite_result.show()
```

```python
for i, result in enumerate(bank_suite_result.results):
    print(i, result.get_header())

results_by_name = {
    result.get_header(): result
    for result in bank_suite_result.results
}

single_value_result = results_by_name["Single Value in Column"]
duplicates_result = results_by_name["Data Duplicates"]
```

```python
print("Single-value result:")
print(single_value_result.value)

print("\nDuplicate-row result:")
print(duplicates_result.value)
```
